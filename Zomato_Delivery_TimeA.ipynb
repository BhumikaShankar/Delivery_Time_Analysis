### Lets the calculate the distance between restaurant and order place using geopy

```python
##!pip install geopy
```

```python
from geopy.distance import geodesic

def calculate_distance(row):
    restaurant = (
        row["Restaurant_latitude"],
        row["Restaurant_longitude"]
    )

    delivery = (
        row["Delivery_location_latitude"],
        row["Delivery_location_longitude"]
    )

    return geodesic(restaurant, delivery).km

df2["Distance"] = df2.apply(calculate_distance, axis=1)
```

### Lets find the outliers using IQR

```python
## Time_Taken
Q1 = df['Time_taken'].quantile(0.25)
Q3 = df['Time_taken'].quantile(0.75)

IQR = Q3-Q1

LB = Q1 - 1.5*IQR
UB = Q3 + 1.5*IQR

outliers = df2[
    (df2['Time_taken']<LB) | (df2['Time_taken']>UB)]

print("No of Outliers: ",len(outliers))
print("IQR:", IQR)
```

## Hypothesis Testing

### Null-hypothesis: The delivery time is same across all the weather six categories(Cloudy, Fog, Sandstorms, Stormy, Sunny, Windy)
### Alternate-hypothesis : The delivery time is higher in atleast one weather category

```python
df2.groupby("Weather_conditions")['Time_taken'].agg(["count","mean","std"]).round(2)
```

### 1.The Pattern shown is shown that: The delivery time is higher when weather is CLOUDY and FOG with an avg time of 29mins.
### 2. The delivery time is less when weather is Sunny with an avg of 22 mins
### 3. The delivery time is fairly same when weather is Stormy, Windy, Sandstorms time of avg 26mins

### ANOVA one way test for proving null-hypothesis with significance value of 0.05

```python
from scipy.stats import f_oneway

from scipy.stats import f_oneway

weather_groups = [
    group["Time_taken"].dropna()
    for _, group in df2.groupby("Weather_conditions")
]

f_stat, p_value = f_oneway(*weather_groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

### The p-value is 0 which is < 0.05 hence null hypothesis is rejected. Therefore, The delivery time is higher in atleast one weather category

## Conclusion

1. **Overall delivery performance:** The mean delivery time was approximately 26 minutes, while the maximum observed delivery time was 54 minutes.

2. **Weather conditions:** Average delivery time differed across weather categories. Cloudy and foggy conditions had the highest average delivery times, at approximately 29 minutes, while sunny conditions had the lowest, at approximately 22 minutes. ANOVA indicated statistically significant differences in mean delivery time across at least some weather categories.

3. **Long delivery times:** Some deliveries took substantially longer than the typical delivery. Traffic density, weather conditions, distance, and multiple deliveries are potential factors associated with delivery time, but this analysis does not establish the causes of individual outliers.

4. **Multiple deliveries and distance:** Regression analysis found statistically significant associations between delivery time, distance, and the number of multiple deliveries. The relationship between distance and delivery time also varied across multiple-delivery groups. These findings suggest that delivery workload deserves further operational investigation.

5. **Order type:** The one-way ANOVA did not find sufficient statistical evidence of differences in mean delivery time across order types at the 5% significance level.

6. **Distance:** Distance had a positive linear correlation with delivery time (Pearson's r ≈ 0.32), but the plot between them indicated a weak linear relationship, which inhibits us from concluding longer distances can create higher delivery time.

7. **Traffic density:** Deliveries under jam traffic conditions had an average delivery time of approximately 31.4 minutes, compared with approximately 21.5 minutes under low-traffic conditions. ANOVA indicated statistically significant differences in mean delivery time across at least some traffic-density categories.

## Limitations

- This is observational data; the findings establish associations, not proof of causation.
- Missing-value removal may introduce bias if the excluded records differ systematically from the retained records.
- The analysis does not establish whether the findings generalise to other cities, periods, or delivery platforms.
- The analysis identifies operational patterns but does not directly measure customer satisfaction, retention, or the financial impact of delivery delays.
